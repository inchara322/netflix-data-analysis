# Netflix Data Cleaning & Preparation

A Python and pandas project that cleans a Netflix titles dataset and exports an analysis-ready file for business reporting.

## Objective

Raw data often contains hidden problems that give wrong results in analysis. The goal of this project was to find and fix those problems so the dataset is reliable for business analytics and reporting.

## Dataset

- **Source:** [add where you got the dataset, for example Kaggle]
- **Size:** 8,790 rows and 10 columns
- **Columns:** `show_id`, `type`, `title`, `director`, `country`, `date_added`, `release_year`, `rating`, `duration`, `listed_in`

## Tools Used

- Python 3.14
- pandas
- Jupyter Notebook in VS Code

## Cleaning Steps

1. **Import and inspect:** loaded the CSV and worked on a copy so the raw data stayed untouched.
2. **Hidden missing values:** `isnull()` reported zero missing values, but the text "Not Given" was used as a placeholder in `director` (2,588 rows) and `country` (287 rows). I replaced it with "Unknown".
3. **Whitespace:** stripped extra spaces from text columns (one title had a hidden trailing space).
4. **Duplicates:** removed 4 duplicate titles that appeared under different `show_id` values.
5. **Dates:** converted `date_added` from text to a real date type and created `year_added` and `month_added`.
6. **Standardization:**
   - Checked that `type` contains only `Movie` and `TV Show`.
   - Created `primary_country` (first country listed) so multi-country titles can be grouped.
   - Created `rating_group` (Kids, Older Kids, Teens, Adults, Not Rated) and kept the original `rating`.
   - Split `duration` into `duration_value` and `duration_unit`.
7. **Export:** validated the result and saved `netflix_cleaned.csv`.

## Results

                          Before   After 
  
| Rows                     8,790   8,786 
| Columns                  10       16 
| Hidden missing values    2,875    0 
| Duplicates               4        0 

New columns: `year_added`, `month_added`, `primary_country`, `rating_group`, `duration_value`, `duration_unit`.

## Key Decision

I kept the 2,588 rows with an unknown director instead of deleting them. Removing them would have discarded about 29% of the data, including useful information such as type, rating, and country. I also kept the original `rating` column and added `rating_group` beside it, because grouping loses detail (for example, TV-14 and PG-13 both become "Teens").
